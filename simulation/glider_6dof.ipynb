### 4. Use a design you exported from the web calculator

In the web app's **Requirements** card, press *“Export design for 6-DOF simulator (.json)”* to get `glider_design.json`, then run the cell below and pick that file when prompted.

```python
import json, glider_6dof as G
from IPython.display import Image, display
try:
    from google.colab import files
    up = files.upload()                       # choose your glider_design.json
    design = json.loads(next(iter(up.values())).decode())
except Exception:
    design = json.load(open('lab_glider_design.json'))

t, hist, glider = G.simulate(glider=G.Glider6DOF.from_design(design),
                             t_max=400, dt=0.02, psi_cmd_deg=25, turn_mode='rudder')
G.plot(t, hist, glider, out='mine.png')
G.animate(t, hist, glider, out='mine.gif', n_frames=120)
display(Image('mine.png'))
display(Image('mine.gif'))
```

*Tip:* change `turn_mode='rudder'` to `'bank'` to try the lateral-mass bank-to-turn, or edit `psi_cmd_deg` to command a different heading.
